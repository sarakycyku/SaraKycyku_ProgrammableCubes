```python
import subprocess, os
os.makedirs("logs", exist_ok=True)

PROBLEM = "Enterprise"
TRIALS = 2000000
SEEDS = [0, 1, 2, 3]

procs = []
for s in SEEDS:
    log = open(f"logs/{PROBLEM}_seed_{s}.log", "w")
    cmd = ["python", "annealing_v3.py", PROBLEM,
           "--trials", str(TRIALS), "--seed", str(s),
           "--t0", "0.5", "--t1", "0.25",
           "--p-mis", "0.7", "--p-near", "0",
           "--tag", f"kaggle_seed_{s}"]
    procs.append((s, subprocess.Popen(cmd, stdout=log, stderr=subprocess.STDOUT), log))

for s, p, log in procs:
    p.wait()
    log.close()
    print("seed", s, "exit code", p.returncode)
```
